# 1. Decision Making 决策分析

## Concept 概念

Decision-making is about choosing the best action under uncertainty，也就是在不确定性环境中做出最优选择。

---

## Key Idea 核心思想

You must think about future decisions first，也就是说在做当前决策之前，必须先考虑未来可能发生的情况。

课程中的关键点是：
“MUST think about what to do on Tuesday first”，说明决策是一个向前看的过程，而不是只看当下。

---

## Method 方法

* Expected Value（期望值）：用来衡量长期平均收益
* Compare different strategies：比较不同方案的期望结果

本质是：选择长期收益最大的策略，而不是短期看起来最好的选择。

---

# 2. Hypothesis Testing 假设检验

## Concept 概念

Hypothesis testing is used to make conclusions about a population based on sample data，也就是通过样本数据对总体做推断。

---

## Intuition 直觉

核心问题是：
“If the assumption is true, how likely is this data?”
也就是如果假设是真的，这个数据出现的概率有多小？

如果非常不可能 → 说明假设可能是错的。

---

## Step-by-step 步骤

1. Hypotheses（提出假设）
   H0：默认成立的假设
   H1：我们想证明的假设

2. Model（选择模型）
   proportion → 二项分布 / 正态近似
   mean → t-test 或 normal

3. Decision Rule（决策规则）
   通常设 α = 0.05

4. Compute（计算）
   计算 test statistic

5. Conclusion（结论）

---

## Sampling Distribution 抽样分布

Sampling distribution 是在原假设成立时，样本统计量的分布，也就是我们用来判断“这个结果正不正常”的参考标准。

可以理解为：
“如果 H0 是真的，我们通常会看到什么样的数据？”

---

## Simulation Example 模拟例子

```python
import numpy as np

n = 100000
size = 400
prob = 0.80

W = np.random.binomial(size, prob, n)
prop = W / size

p_value = np.mean(prop <= 0.73)
p_value
```

这个结果就是 p-value，也就是在 H0 成立的前提下，观察到这么极端结果的概率。

---

## Errors 错误类型

* Type I Error：reject true H0，也就是错杀（假阳性）
* Type II Error：fail to reject false H0，也就是放过错误（假阴性）

---

## Conclusion Rule 结论规则

当 p-value < 0.05 时，说明这个结果在 H0 下非常不可能，因此我们 reject H0。

可以理解为：
数据太“异常”，不能再相信原假设。

# 3. Confidence Interval 置信区间

## Concept 概念

Confidence interval 是一个区间，用来估计真实参数的位置。

---

## Important Clarification 重要理解

95% CI does NOT mean there is a 95% probability that the parameter lies in the interval。
正确理解是：如果重复抽样，95% 的区间会包含真实参数。

换句话说：
置信区间描述的是方法的可靠性，而不是某一个区间的概率。

---

# 4. Simulation 模拟

## Concept 概念

Simulation uses random sampling to approximate complex problems，也就是通过随机生成数据来逼近真实结果。

---

## Why Simulation 为什么用

* Real systems are complex，现实系统太复杂
* No closed-form solution，没有简单公式可以直接计算

---

## Steps 步骤

1. Generate random variables（生成随机变量）
2. Compute result（计算结果）
3. Repeat many times（重复多次）

本质是用大量重复来逼近真实分布。

---

## Example 课程案例

```python
N = 1000

L = np.random.uniform(5040, 6860, N)
M = np.random.normal(3000, 1000, N)

P = np.random.choice([20,18.5,16.5,15], N, p=[0.25,0.35,0.30,0.10])

R = P * M
C = L + 3995 + 11*M

E = R - C

np.mean(E), np.std(E)
```

---

## Key Insight 核心理解

Simulation approximates unknown distributions，也就是说模拟是在用计算机逼近我们无法直接求解的概率分布。

# 5. Chi-Square Test 卡方检验

## Concept 概念

Chi-square test is used to test the relationship between two categorical variables，也就是检验两个分类变量是否有关联。

---

## Hypothesis 假设

H0：No association（没有关系）
H1：There is association（有关系）

---

## Formula 公式

χ² = Σ (Observed - Expected)² / Expected

本质是比较“观察值”和“理论值”的差距。

---

## Decision 决策

当 p-value < 0.05 时，说明差异显著，可以认为变量之间存在关系。

---

# Final Summary 总结

## Big Picture

Decision-making 是用来选择最优策略的工具，
Hypothesis testing 用来验证假设是否合理，
Simulation 用来处理复杂且无法解析的问题，
Statistics 则为整个决策过程提供理论支持。

整体来看，这些方法的核心都是：
用数据来支持理性决策，而不是依赖直觉。
