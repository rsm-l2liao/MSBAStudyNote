# SQL 基础

## 字段命名规范

* 小写
* 使用下划线
* 单数形式




## SELECT

```sql
SELECT * FROM table;
```

要点 `*` = wildcard（所有列）





## JOIN

### INNER JOIN

```sql
SELECT c.name, l.name
FROM countries AS c
INNER JOIN languages AS l
USING(code);
```

要点：
* 只保留两表都有的数据

### LEFT JOIN

```sql
SELECT *
FROM A
LEFT JOIN B
ON A.id = B.id;
```

要点：

* 保留左表全部
* 右表没有 → NULL


## Alias

```sql
SELECT name AS first_name
FROM employees;
```





## DISTINCT

```sql
SELECT DISTINCT year
FROM employees;
```





## VIEW

```sql
CREATE VIEW my_view AS
SELECT name FROM table;
```





## 多表 JOIN

```sql
SELECT name, fertility_rate
FROM countries AS c
INNER JOIN populations AS p
ON c.code = p.country_code;
```





## UNION

```sql
SELECT * FROM A
UNION
SELECT * FROM B;
```

要点：

* UNION：去重
* UNION ALL：不去重
