
# SQL Learning Notes

# 1. Basic SELECT（基础查询）

## Concept（概念）

**EN**
Retrieve data from a table using SELECT.

**中文**
使用 SELECT 从数据库中获取数据。





## Syntax（语法）

```sql
SELECT column1, column2
FROM table_name;
```



## Filtering（条件筛选）

```sql
SELECT *
FROM table
WHERE age > 18 AND country = 'US';
```

Use WHERE to filter rows.
使用 WHERE 筛选数据。



## Sorting（排序）

```sql
SELECT *
FROM table
ORDER BY salary DESC;
```





# 2. CASE WHEN（条件逻辑）

## Concept（概念）

Implements conditional logic (if-else).
SQL中的条件判断（类似 if-else）。


```sql
SELECT name,
CASE 
    WHEN score >= 90 THEN 'A'
    WHEN score >= 80 THEN 'B'
    ELSE 'C'
END AS grade
FROM students;
```





# 3. JOIN（表连接）

## Concept（概念）

Combine data from multiple tables.
将多个表的数据连接起来。



## INNER JOIN

```sql
SELECT A.id, B.name
FROM tableA A
INNER JOIN tableB B
ON A.id = B.id;
```

只保留匹配数据



## LEFT JOIN

```sql
SELECT A.id, B.name
FROM tableA A
LEFT JOIN tableB B
ON A.id = B.id;
```

保留左表全部数据



## Key Insight（重点）

| 类型         | 含义   |
| ---------- | ---- |
| INNER JOIN | 交集   |
| LEFT JOIN  | 保留左表 |





# 4. Aggregation（聚合）

## Concept（概念）

Summarize data across multiple rows.
对多行数据进行汇总。



```sql
SELECT category, COUNT(*), AVG(price)
FROM products
GROUP BY category;
```



## WHERE vs HAVING

| 条件     | 作用  |
| ------ | --- |
| WHERE  | 聚合前 |
| HAVING | 聚合后 |



```sql
SELECT category, AVG(price)
FROM products
GROUP BY category
HAVING AVG(price) > 100;
```





# 5. Subquery & CTE（子查询与CTE）

## Subquery（子查询）

```sql
SELECT *
FROM employees
WHERE salary > (
    SELECT AVG(salary) FROM employees
);
```



## CTE（Common Table Expression）

```sql
WITH temp AS (
    SELECT category, AVG(price) AS avg_price
    FROM products
    GROUP BY category
)
SELECT *
FROM temp;
```





# 6. Window Functions（窗口函数）

## Concept（概念）

Perform calculations across rows without reducing row count.
在不减少行数的情况下进行统计计算。



## 基本写法

```sql
SELECT name, category, price,
AVG(price) OVER (PARTITION BY category) AS avg_price
FROM products;
```


## PARTITION BY（分组）

```sql
OVER(PARTITION BY category)
```
类似 GROUP BY，但不会减少行



## ORDER BY（排序）

```sql
OVER(PARTITION BY id ORDER BY date)
```
定义窗口内顺序



## Running Total（累计和）

```sql
SELECT product_id, date, quantity,
SUM(quantity) OVER (
    PARTITION BY product_id
    ORDER BY date
) AS running_total
FROM sales;
```


## Window Frame（窗口范围）

```sql
ROWS BETWEEN 1 PRECEDING AND CURRENT ROW
```

控制计算范围



## 常见函数

```sql
ROW_NUMBER()
RANK()
DENSE_RANK()
LAG()
LEAD()
```


## Key Insight（核心区别）

| 类型       | 行数变化 |
| -------- | ---- |
| GROUP BY | 减少   |
| WINDOW   | 不变   |





# 7. Recursive CTE（递归查询）

## Concept（概念）

**EN**
Recursion in SQL = looping over results.

**中文**
递归SQL = 在SQL中做循环。



## 基本结构

```sql
WITH RECURSIVE nums AS (
    SELECT 1 AS n
    UNION ALL
    SELECT n + 1
    FROM nums
    WHERE n < 5
)
SELECT * FROM nums;
```



## 执行逻辑

1. 初始查询（base case）
2. 递归查询（recursive step）
3. 循环直到停止


## 应用场景

* 组织结构（员工层级）
* 图/路径问题
* 产品结构（BOM）





# 8. 高级理解（重要）

## Window vs Group

Window functions keep all rows.
窗口函数不会减少数据行。



## Recursion vs Loop

Recursive CTE behaves like a loop.
递归CTE本质就是循环。




# Final Summary

* SELECT: retrieve data 查询数据
* JOIN: combine tables 连接表
* GROUP BY: aggregation 聚合
* WINDOW: advanced analytics 高级分析
* RECURSIVE: hierarchical queries 层级查询
