## 1. Working with Strings（字符串）

```python
# 字符串定义
a = 'hello'
b = "hello"
c = """multi-line
string"""

# 常用方法
text = "Hello World"
print(text.lower())
print(text.upper())
print(text.replace("World", "Python"))
```

要点：
* 单引号 / 双引号 / 三引号（多行）
* 常见方法：`.lower()`, `.upper()`, `.replace()`



## 2. List（列表）

```python
a_list = [1, 2, 3, 4, 5]

# 索引
print(a_list[0])     # 第一个
print(a_list[-1])    # 最后一个

# 切片
print(a_list[0:3])
print(a_list[::2])

# 添加元素
a_list.append(6)
print(a_list)
```

要点：
* 可存多种类型（str, int, float, bool）
* 从 index 0 开始
* `append()` 会修改原列表





## 3. Dictionary（字典）

```python
products = {
    "A": 10,
    "B": 20
}

# 访问
print(products["A"])

# 添加
products["C"] = 30

# 遍历
for k, v in products.items():
    print(k, v)
```

要点：
* key-value 结构
* `.items()` 常用于循环




## 4. Sets & Tuples

```python
# Set（无序、无重复）
s = {1, 2, 3}

# Tuple（不可变）
t = (1, 2, 3)
```

对比：
| 类型    | 可变 | 有序 | 可重复    |
| ----- | -- | -- | ------ |
| List  | ✅  | ✅  | ✅      |
| Dict  | ✅  | ✅  | key不重复 |
| Set   | ✅  | ❌  | ❌      |
| Tuple | ❌  | ✅  | ✅      |



## 5. 条件语句

```python
x = 10

if x > 10:
    print("big")
elif x == 10:
    print("equal")
else:
    print("small")
```

运算符：

* `== != > < >= <=`





## 6. 循环

### For loop

```python
prices = [9.9, 8.9, 5.5]

for p in prices:
    print(p)
```

### While loop

```python
count = 0
while count < 5:
    print(count)
    count += 1
```

while：条件满足就一直运行



## 7. Range

```python
for i in range(1, 5):
    print(i)
```

要点：

* start：包含
* end：不包含




## 8. 运算符

```python
print(4 + 2)
print(4 * 2)
print(7 % 4)
print(3 ** 2)
```

要点：

* `%` 取余
* `**` 幂运算




##  9. 常用函数

```python
nums = [1, 2, 3]

print(len(nums))
print(max(nums))
print(min(nums))
print(sum(nums))
print(sorted(nums))
```

常见函数：
`print(), len(), max(), min(), sum(), sorted()`





## 10. Iterators（迭代器）

```python
my_list = [1, 2, 3]

it = iter(my_list)

print(next(it))
print(next(it))
```

要点：

* `iter()` 创建迭代器
* `next()` 取值
* 用完会报错 StopIteration





## 11. Functions & Docstring

```python
def add(a, b):
    """Return sum of a and b"""
    return a + b

print(add(2, 3))
```

要点：

* Docstring 可被程序读取
* `*args`, `**kwargs` 支持多参数



## 12. Error Handling

```python
try:
    x = int("abc")
except:
    print("error")
```

要点：

* try-except：防止程序崩溃
* raise：主动报错
